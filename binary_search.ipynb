## 34 Find First and Last Position of Element in Sorted Array

```python
from typing import List

class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:

        def search(firstBound: bool) -> int:
            left = 0
            right = len(nums) - 1
            bound = -1
            
            # Loop
            while left <= right:
                # Calculate the mid index 
                mid = (left + right) // 2

                # Check the value of index if it matches the target 
                if nums[mid] == target:
                    bound = mid
                    if firstBound:
                        right = mid - 1

                    else:
                        left = mid + 1

                elif nums[mid] < target:
                    left = mid + 1

                elif nums[mid] > target:
                    right = mid - 1
            
            return bound
        
        # Begin
        firstResult= search(firstBound=True)
        if firstResult == -1:
            # No matching values
            return [-1, -1]
        
        secondResult = search(firstBound=False)

        # Return final
        return [firstResult, secondResult]
```

## 74. Search a 2D Matrix

```python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        # Get the dimensions of matrix 
        m = len(matrix)    #Number of rows
        n = len(matrix[0]) #Number of columns

        # Get the left and right pointers 
        left = 0 
        right = (m*n) - 1

        # Loop
        while left <= right:
            # Calculate the  mid index 
            mid = (left + right) // 2

            # Get the mid value
            mid_value = matrix[mid // m][mid % n]

            if mid_value == target:
                return True
            
            elif mid_value < target:
                left = mid + 1

            elif mid_value > target:
                right = mid - 1
        
        return False
```

## 81. Search in Rotated Sorted Array II

```python
class Solution:
    def search(self, nums: List[int], target: int) -> bool:
        left = 0 
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return True
            
            elif nums[left] == nums[mid] == nums[right]:
                left += 1
                right -= 1 
            
            # Search which side sorted 
            elif nums[mid] <= nums[right]:
                # right side sorted 
                # Search the target 
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                
                else:
                    right = mid - 1 
            
            else:
                # Left side sorted 
                # Search 
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                
                else:
                    left = mid + 1

        return False
```

## Find Peak Elements 

```python
from typing import List

class Solution:
    def findPeakElement(self, nums: List[int]) -> int:
        # Set the pointers 
        left = 0
        right = len(nums) - 1

        while left < right:
            # Calculate mid value 
            mid = (left + right) // 2 

            # Compare values 
            if nums[mid] > nums[mid + 1]:
                right = mid
            
            elif nums[mid] < nums[mid + 1]:
                left = mid + 1

        return left
```
