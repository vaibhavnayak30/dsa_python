```python
class Solution:
    def minSubArrayLen(self, nums: List[int], target: int) -> int:
        # Initialise variables 
        left = 0 
        cur_sum = 0
        answer = float('inf')

        # Loop through each element 
        for right in range(len(nums)):
            # Calculate current sum 
            cur_sum += nums[right]

            while cur_sum >= target:
                answer = min(answer, right - left + 1)
                
                cur_sum -= nums[left]
                left += 1
        
        return 0 if answer == float('inf') else answer
```

```python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        # Initiate variables 
        answer = [1] * len(nums)

        # Calculate prefix 
        prefix = 1
        for i in range(len(nums)):
            answer[i] = prefix
            prefix *= nums[i]

        # Calculate suffix 
        suffix = 1
        for i in range(len(nums) - 1, -1, -1):
            answer[i] *= suffix
            suffix *= nums[i]
        
        return answer
```

```python
import heapq 

class Solution:
    def minMeetingRooms(self, intervals: List[List[int]]) -> int:
        # Sort the intervals 
        intervals.sort(key= lambda x: x[0])

        # Initialise min heap 
        free_room = []
        count = 1

        # Add first 
        heapq.heappush(free_room, intervals[0][1])

        # loop
        for meeting in intervals[1:]:
            if meeting[0] >= free_room[0]:
                heapq.heappop(free_room)
                heapq.heappush(free_room, meeting[1])
            
            else:
                heapq.heappush(free_room, meeting[1])
                count += 1
        
        return count
```

```python
class Solution:
    def canAttendMeetings(self, intervals: List[List[int]]) -> bool: 
        # If empty intervals 
        if len(intervals) == 0:
            return True
        
        # Sort the intervals 
        intervals.sort(key= lambda x: x[0])

        # Person joined the first meeting
        getting_freed= intervals[0][1]

        # Loop and check 
        for meeting in intervals[1:]:
            # If person next meeting is before current meeting ending time
            if meeting[0] < getting_freed:
                return False
            
            getting_freed = meeting[1]

        return True
```
