## Longest Palindromic Substring

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        result = ''

        # Write expand function 
        def expand(left, right):
            while left >= 0 and right < len(s) and s[left] == s[right]:
                left -= 1
                right += 1

            return s[left + 1 : right]

        # Loop
        for idx in range(len(s)):
            odd = expand(left=idx, right= idx)
            even = expand(left=idx, right= idx+1)

            longer = odd if len(odd) > len(even) else even

            if len(longer) > len(result):
                result = longer
            
        return result
```

## Container With Most Water

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        # Initialise the pointers 
        left = 0
        right = len(height) - 1
        result = 0

        # Loop
        while left < right:
            # Calculate the current area 
            curr_water_content = min(height[left], height[right]) * (right - left)

            # Get the max 
            result = max(result, curr_water_content)

            # Shift the pointers 
            if height[left] < height[right]:
                left += 1
            
            else:
                right -= 1

        return result
```
